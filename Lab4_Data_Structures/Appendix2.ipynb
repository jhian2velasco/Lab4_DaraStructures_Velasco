```python
from collections import defaultdict

LAST_NAME = "Velasco"
SEED_NUM = 4
FAVORITE_ARTIST = "Maria Deborah"

students = [f"{LAST_NAME}_S{i}" for i in range(1, 4)]

subjects = [
    "PHYSICAL",
    "SPECIAL",
    "STATUS",
    "DEFENSE",
    "SPEED",
]

print("Students:", students)
print("Subjects:", subjects)
```

```python
artist_positions = [i + 1 for i in range(len(FAVORITE_ARTIST))]

records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):

        score = ((SEED_NUM *
                 artist_positions[(i + j) %
                 len(artist_positions)])
                 % 51) + 50

        records.append({
            "Student": student,
            "Subject": subject,
            "Score": score
        })

# Repeated record
records.append(records[0].copy())

# Incomplete record
records.append({
    "Student": students[2],
    "Subject": "PHYSICAL",
    "Score": None
})

print("GENERATED PERFORMANCE RECORDS\n")

for record in records:
    print(record)
```

```python
def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 70:
        return "GOOD"
    elif score >= 60:
        return "FAIR"
    else:
        return "POOR"

for record in records:
    record["subjects"] = performance_category(record["Score"])

print("PERFORMANCE CATEGORIES ADDED")
```

```python
print("ORGANIZED STUDENT/SUBJECT DATA\n")

for record in records:
    print(
        f"{record['Student']} | "
        f"{record['Subject']} | "
        f"{record['Score']} | "
        f"{record['subjects']}"
    )
```

```python
counter = defaultdict(int)

for record in records:
    key = (record["Student"], record["Subject"])
    counter[key] += 1

repeated_records = []

for key, count in counter.items():
    if count > 1:
        repeated_records.append(key)

print("REPEATED RECORDS\n")

for record in repeated_records:
    print(record)
```

```python
incomplete_records = []

for record in records:
    if record["Score"] is None:
        incomplete_records.append(record)

print("INCOMPLETE RECORDS\n")

for record in incomplete_records:
    print(record)
```

```python
student_summary = defaultdict(list)

for record in records:
    if record["Score"] is not None:
        student_summary[record["Student"]].append(record["Score"])

print("STUDENT SUMMARIES\n")

for student, scores in student_summary.items():

    average = sum(scores) / len(scores)

    print(f"\n{student}")
    print("Average:", round(average, 2))
    print("Highest:", max(scores))
    print("Lowest :", min(scores))
```

```python
subject_summary = defaultdict(list)

for record in records:
    if record["Score"] is not None:
        subject_summary[record["Subject"]].append(record["Score"])

print("SUBJECT SUMMARIES\n")

for subject, scores in subject_summary.items():

    average = sum(scores) / len(scores)

    print(f"{subject}: Average Score = {average:.2f}")
```

```python
ordered_results = sorted(
    [r for r in records if r["Score"] is not None],
    key=lambda x: x["Score"],
    reverse=True
)

print("ORDERED RESULTS\n")

for record in ordered_results:
    print(
        f"{record['Student']} | "
        f"{record['Subject']} | "
        f"{record['Score']} | "
        f"{record['subjects']}"
    )
```

```python
valid_scores = []

for record in records:
    if record["Score"] is not None:
        valid_scores.append(record["Score"])

overall_average = sum(valid_scores) / len(valid_scores)

print("=" * 50)
print("FINAL STUDENT PERFORMANCE REPORT")
print("=" * 50)

print("Total Records      :", len(records))
print("Valid Records      :", len(valid_scores))
print("Repeated Records   :", len(repeated_records))
print("Incomplete Records :", len(incomplete_records))
print("Overall Average    :", round(overall_average, 2))
```
